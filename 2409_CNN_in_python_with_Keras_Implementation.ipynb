```python
from keras.datasets import mnist
import numpy as np
import matplotlib.pyplot as plt
%matplotlib inline
```

```python
# Load the data
(train_X,train_Y), (test_X,test_Y) = mnist.load_data()
print('Training data shape : ', train_X.shape, train_Y.shape)
print('Testing data shape : ', test_X.shape, test_Y.shape)
```

```python
# Find the unique numbers from the train labels
classes = np.unique(train_Y)
nClasses = len(classes)
print('Total number of outputs : ', nClasses)
print('Output classes : ', classes)
```

```python
mnistDict = {
    0:"Digit 0",
    1:"Digit 1",
    2:"Digit 2",
    3:"Digit 3",
    4:"Digit 4",
    5:"Digit 5",
    6:"Digit 6",
    7:"Digit 7",
    8:"Digit 8",
    9:"Digit 9",
}
```

```python
# Display the image in dataset
plt.figure(figsize=[5,5])
plt.subplot(121)
plt.imshow(train_X[0,:,:], cmap='gray')
category = mnistDict[train_Y[0]]
plt.title("Label : {}".format(train_Y[0]))
print("Category: ", category)
```

```python
train_X = train_X.reshape(-1, 28,28, 1)
test_X = test_X.reshape(-1, 28,28, 1)
train_X.shape, test_X.shape
```

```python
train_X = train_X.astype('float32')
test_X = test_X.astype('float32')
train_X = train_X / 255.
test_X = test_X / 255.
```

```python
from keras.utils import to_categorical

# Change the labels from categorical to one-hot encoding
train_Y_one_hot = to_categorical(train_Y)
test_Y_one_hot = to_categorical(test_Y)

# Display the change for category label using one-hot encoding
print('Original label:', train_Y[0])
print('After conversion to one-hot:', train_Y_one_hot[0])
```

```python
from sklearn.model_selection import train_test_split
train_X,valid_X,train_label,valid_label = train_test_split(train_X, train_Y_one_hot, test_size=0.2, random_state=1)
```

```python
print("Shape of the Train and Validation Dataset: ")
train_X.shape,valid_X.shape,train_label.shape,valid_label.shape
```

```python
import keras
from keras.layers import Input
from keras.models import Sequential,Model
from keras.layers import Dense, Dropout, Flatten
from keras.layers import Conv2D, MaxPooling2D
from keras.layers import BatchNormalization
from keras.layers import LeakyReLU
```

```python
BATCH_SIZE = 64 # 128, 256
EPOCHS = 20
NUMBER_OF_CLASSES = 10
```

```python
digit_model = Sequential()
digit_model.add(Conv2D(32, kernel_size=(3, 3),activation='linear',input_shape=(28,28,1),padding='same'))
digit_model.add(LeakyReLU(alpha=0.1))
digit_model.add(MaxPooling2D((2, 2),padding='same'))
digit_model.add(Conv2D(64, (3, 3), activation='linear',padding='same'))
digit_model.add(LeakyReLU(alpha=0.1))
digit_model.add(MaxPooling2D(pool_size=(2, 2),padding='same'))
digit_model.add(Conv2D(128, (3, 3), activation='linear',padding='same'))
digit_model.add(LeakyReLU(alpha=0.1))
digit_model.add(MaxPooling2D(pool_size=(2, 2),padding='same'))
digit_model.add(Flatten())
digit_model.add(Dense(128, activation='linear'))
digit_model.add(LeakyReLU(alpha=0.1))
digit_model.add(Dense(nClasses, activation='softmax'))
```

```python
digit_model.compile(loss=keras.losses.categorical_crossentropy, optimizer=keras.optimizers.Adam(),metrics=['accuracy'])
```

```python
digit_model.summary()
```

```python
Digit_train = digit_model.fit(train_X, train_label, batch_size=BATCH_SIZE,epochs=EPOCHS,verbose=1,validation_data=(valid_X, valid_label))
```

# Model Evaluation

```python
test_eval = digit_model.evaluate(test_X, test_Y_one_hot, verbose=1)
print('Test loss:', test_eval[0])
print('Test accuracy:', test_eval[1])
```

```python
accuracy = Digit_train.history['accuracy']
val_accuracy = Digit_train.history['val_accuracy']
loss = Digit_train.history['loss']
val_loss = Digit_train.history['val_loss']
epochs = range(len(accuracy))

plt.plot(epochs, accuracy, 'bo', label='Training accuracy')
plt.plot(epochs, val_accuracy, 'b', label='Validation accuracy')
plt.title('Training and validation accuracy')
plt.legend()
plt.figure()

plt.plot(epochs, loss, 'bo', label='Training loss')
plt.plot(epochs, val_loss, 'b', label='Validation loss')
plt.title('Training and validation loss')
plt.legend()
plt.show()
```

```python
digit_model = Sequential()
digit_model.add(Conv2D(32, kernel_size=(3, 3),activation='linear',padding='same',input_shape=(28,28,1)))
digit_model.add(LeakyReLU(alpha=0.1))
digit_model.add(MaxPooling2D((2, 2),padding='same'))
digit_model.add(Dropout(0.25))
digit_model.add(Conv2D(64, (3, 3), activation='linear',padding='same'))
digit_model.add(LeakyReLU(alpha=0.1))
digit_model.add(MaxPooling2D(pool_size=(2, 2),padding='same'))
digit_model.add(Dropout(0.25))
digit_model.add(Conv2D(128, (3, 3), activation='linear',padding='same'))
digit_model.add(LeakyReLU(alpha=0.1))
digit_model.add(MaxPooling2D(pool_size=(2, 2),padding='same'))
digit_model.add(Dropout(0.4))
digit_model.add(Flatten())
digit_model.add(Dense(128, activation='linear'))
digit_model.add(LeakyReLU(alpha=0.1))
digit_model.add(Dropout(0.3))
digit_model.add(Dense(nClasses, activation='softmax'))
```

```python
digit_model.summary()
```

```python
digit_model.compile(loss=keras.losses.categorical_crossentropy, optimizer=keras.optimizers.Adam(),metrics=['accuracy'])
```

```python
Digit_train_dropout = digit_model.fit(train_X, train_label, batch_size=BATCH_SIZE,epochs=EPOCHS,verbose=1,validation_data=(valid_X, valid_label))
```
